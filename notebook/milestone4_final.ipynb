```python
results = []

emails = [
    "Please send the report by EOD",
    "Schedule a meeting tomorrow at 10 AM",
    "Thanks for your help"
]

for e in emails:
    res = app.invoke({"email": e})
    results.append(res)

results
```

```python
import pandas as pd

df = pd.DataFrame(results)
df
```
