```python
DANGEROUS_TOOLS =  ["send_email","create_calendar_invite"]
```

```python
def agent_decide_action(body):
    if "meeting" in body.lower():
        return "create_calendar_invite"
    if "reply" in body.lower():
        return "send_email"
    return "read_calendar"
```

```python
def hitl_checkpoint(action):

    if action in DANGEROUS_TOOLS:
       return "Wait_for_Human"
    return "SAFE"
```

```python
def human_decision():
    return "Deny"  # Simulated human decision
```

```python
def execute_tool(action):
    return f"Executed action: {action}"
```

```python
def run_agent(body):
    print("Email received:",body)
    action = agent_decide_action(body)
    print("Decided action:",action)

    status= hitl_checkpoint(action)
    print("HITL Status:",status)

    if status == "Wait_for_Human":
        decision = human_decision()
        print("Human decision:",decision)
        if decision == "Approve":
            return execute_tool(action)
        elif decision == "Deny":
            return "Action denied by human."
        elif decision =="Edit":
            return "Action edited by human."
    return execute_tool(action)
```

```python
email="Please schedule a meeting with the client."
result= run_agent(email)
print(result)
```
