```python
def triage_rule_v2(text):
    text = text.lower()
    
    if any(keyword in text for keyword in ["refund", "urgent", "password", "invoice"]):
        return "notify_human"
    
    if any(keyword in text for keyword in ["newsletter", "promotion", "sale"]):
        return "ignore"
    
    return "respond_or_act"
```
