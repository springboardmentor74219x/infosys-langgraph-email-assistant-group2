```python
import pandas as pd

# Load the email dataset
df = pd.read_csv("../data/sample_emails_with_triage_200.csv")

df.head()
```

```python
# Actions that require Human-in-the-Loop approval
DANGEROUS_TOOLS = ["send_email", "create_calendar_invite"]
```

```python
def agent_decide_action(body: str) -> str:
    body = body.lower()

    if "meeting" in body:
        return "create_calendar_invite"
    elif "reply" in body:
        return "send_email"
    else:
        return "read_calendar"
```

```python
def hitl_checkpoint(action: str) -> str:
    if action in DANGEROUS_TOOLS:
        return "WAIT_FOR_HUMAN"
    return "SAFE"
```

```python
def human_decision() -> str:
    # Change to "Deny" or "Edit" to test behavior
    return "Approve"
```

```python
def execute_tool(action: str) -> str:
    return f"Executed action: {action}"
```

```python
def run_agent(body: str) -> str:
    print("Email received:", body)

    action = agent_decide_action(body)
    print("Agent decided action:", action)

    status = hitl_checkpoint(action)
    print("HITL status:", status)

    if status == "WAIT_FOR_HUMAN":
        decision = human_decision()
        print("Human decision:", decision)

        if decision == "Approve":
            return execute_tool(action)
        elif decision == "Deny":
            return "Action denied by human"
        elif decision == "Edit":
            return "Action edited by human"

    return execute_tool(action)
```

```python
email = "Please schedule a meeting with the client"
run_agent(email)
```
