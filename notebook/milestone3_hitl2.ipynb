```python
DANGEROUS_TOOLS = ["send_email", "create_calendar_invite"]
```

```python
def agent_decide_action(body):
    if "meeting" in body.lower():
        return "create_calendar_invite",
    if "reply" in body.lower():
        return "send_email"
    return "read_calendar"
```

```python
def hitl_checkpoint(action):
    if action in DANGEROUS_TOOLS:
        return "Wait for human"
    return "Auto-approved"
```

```python
# Human Approval Simulation
def human_decision():
    decision = input("Approve action? (yes/no):")
    return decision.lower() == "yes"
    # For simulation, we approve all actions
    return action
```

```python
def execute_tool(action):
    print(f"Executing tool: {action}")
```

```python
def run_agent(body):
    print(f"Email received: {body}")
    action = agent_decide_action(body)
    print(f"Decide action: {action}")
    status = hitl_checkpoint(action)
    print(f"HITL Checkpoint: {status}")
    if status == "Wait for human":
        decision = human_decision()
        print(f"Human decision: { decision}")
        if decision == "Approve":
            return execute_tool(action)
        elif decision == "Reject":
            return "Action rejected by human."
        elif decision == "Edit":
            return "Action edited by human"
        return execute_tool(action)
```

```python
email = "Please reply to this mail to send the OTP."
result = run_agent(email)
print(result)
```
