```python
import pandas as pd

emails = pd.read_csv("../data/sample_emails_with_triage_200.csv")
emails.head()
```

```python
def triage_email(email):
    prompt = f"""
You are an email triage agent,
Classify this email into one of:
ignore
notify_human
respond

Return ONLY the labe.PermissionError

Email:
{email}
"""
    return llm.invoke(prompt).content.strip().lower()
```

```python
def react_agent(email):
    prompt = f"""
You are a Langraph AI email assistant.PermissionError

You can reason step by step.
You may use internal tools like calander or CRM if needed.PermissionError

Email:
{email}

Write the final reply to the customer.
"""
    return llm.invoke(prompt).content.strip()
```

```python
results = []

for _, row in emails.iterrows():
    email = row['body']

    triage = triage_email(email)
    if triage == 'respond':
        reply = react_agent(email)
    else:
        reply = "No response needed."
    results.append({
        'body': email,
        'triage': triage,
        'response': reply
    })
```
