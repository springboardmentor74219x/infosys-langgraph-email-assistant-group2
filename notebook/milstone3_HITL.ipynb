```python
import pandas as pd
df = pd.read_csv("../data/sample_emails_with_triage_200.csv")
df.head()
```

```python
def email_assistant(email_text):
    text = email_text.lower()
    
    urgent_keywords = [
        "urgent", "overdue", "security", "breach", "declined", 
        "failed", "immediate", "server", "crash", "alert"
    ]
    ignore_keywords = [
        "newsletter", "automatic reply", "subscription", "deal", 
        "offer", "tracking", "system generated", "no reply", "marketing"
    ]
    if any(word in text for word in urgent_keywords):
        return "notify", "urgent"

    elif any(word in text for word in ignore_keywords):
        return "ignore", "neutral"
    
    elif "thank you" in text or "congratulations" in text:
        return "respond", "polite"

    else :
        return "respond", "neutral"
```

```python
DANGEROUS_TOOL=["respond"]
```

```python
def hitl_checkpoint(action) :
    if action in DANGEROUS_TOOL:
        return "WAIT_FOR_HUMAN"
    return "SAFE"
```

```python
def human_decision():
    return "Approve"
```

```python
results = []

for _, row in df.sample(10).iterrows():
    action, tone = email_assistant(row["body"])
    status = hitl_checkpoint(action)
    if status == "WAIT_FOR_HUMAN":
        approved = human_decision()
        final_action = action if approved else "blocked"
    else:
        final_action = action
    results.append({
        "email_id": row["body"][:50],
        "ai_action": action,
        "final_action": final_action,
        "hitl_status": status
    })
pd.DataFrame(results)
```

```python
milestone3_hitl_output = pd.DataFrame(results)

output_path = "../data/milestone3_hitl_output_NitishRM.csv"
milestone3_hitl_output.to_csv(output_path, index=False)

print("Milestone 3 HITL output saved at:", output_path)
```
