```python

import pandas as pd


data = {
    'id': [1, 2, 3, 4, 5],
    'sender': [
        'alerts@bank.com', 'alerts@bank.com', 'no-reply@service.com',
        'sales@shop.com', 'no-reply@service.com'
    ],
    'subject': [
        'Password Reset Request', "Congratulations! You've Won", 
        'Promotion: Big Sale', 'Monthly Report', 'Survey'
    ],
    'body': [
        'Reminder: The client meeting is scheduled at 1...',
        'Your invoice of INR 25515.09 is due on 2025-12...',
        'Reminder: The client meeting is scheduled at 1...',
        'Hello team, please find the attached weekly re...',
        'Hello team, please find the attached weekly re...'
    ],
    'priority': ['low', 'low', 'low', 'medium', 'low'],
    'triage_label': ['notify_human', 'respond', 'ignore', 'respond', 'respond']
}
```

```python

df = pd.DataFrame(data)
```

```python

def generate_response(triage):
    if triage == 'notify_human':
        return "This email is important. Please notify or escalate to the human immediately."
    elif triage == 'respond':
        return "Please respond to this email and take the required action."
    elif triage == 'ignore':
        return "No action is required for this email."
    else:
        return "Review the email and take appropriate action."
```

```python
df['ideal_response'] = df['triage_label'].apply(generate_response)

df[['triage_label', 'ideal_response']].head()
```
