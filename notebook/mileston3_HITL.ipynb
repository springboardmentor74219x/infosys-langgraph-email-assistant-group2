```python
import pandas as pd
df = pd.read_csv("../data/sample_emails_with_triage_200.csv")
df.head()
                 
```

```python
#  Email Assistant Logic
def email_assistant(email_text):
    text = email_text.lower()
    if "urgent" in text  or "deadline" in text:
        return "notify", "urgent"
    elif "thank you" in text or "thanks" in text:
        return "ignore", "polite"
    else :
        return "respond", "neutral"
    
```

```python
# define Dangerous Actions
dangerous_actions = ["respond"]
```

```python
# HITl Logic
def hitl_check(action):
    if action in dangerous_actions:
        return "Wait for human review"
    return "Auto approved"
```

```python
# Simulate Human Approval
def human_decision():
    descision = input("Approve action? (yes/no):")
    return descision.lower() == "yes"
```

```python
# 
Results = []

for _, row in df.sample(10).iterrows():
    action, tone = email_assistant(row["body"])
    
    status = hitl_check(action)
    
    if status == "Wait for human":
        approved = human_decision()
        final_action = action if approved else "blocked"
    else:
        final_action = action

    Results.append({
        "email_text": row["body"] [:50],
        "predicted_action": action,
        "predicted_tone": tone,
        "hitl_status": status,
        "final_action": final_action
    })

results_df = pd.DataFrame(Results)
results_df
```

## Save the Milestone3 output file 

```python

results_df.to_csv(
"../data/milestone3_hitl_output_syedumairali.csv",
index=False
)
print("Results saved to milestone3_hitl_output_syedumairali.csv")
```
