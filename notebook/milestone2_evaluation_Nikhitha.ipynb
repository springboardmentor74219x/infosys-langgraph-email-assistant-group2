```python
def email_assistant(text):
    t = str(text).lower()
    
    if any(k in t for k in ['invoice','payment','due','meeting','submit','confirm']):
        action = 'respond'
        tone = 'polite'
    elif any(k in t for k in ['password','security','alert','unauthorized','suspicious']):
        action = 'notify'
        tone = 'urgent'
    else:
        action = 'ignore'
        tone = 'neutral'
        
    return action, tone
```

```python
df[['predicted_action','predicted_tone']] = df['email_text'].apply(
    lambda x: pd.Series(email_assistant(x))
)

df.head()
```

```python
def add_ground_truth(text):
    text = str(text).lower()

    if any(w in text for w in ["submit", "deadline", "meeting", "reminder", "asap", "eod"]):
        return "respond", "urgent"
    elif any(w in text for w in ["thank you", "thanks"]):
        return "ignore", "polite"
    elif any(w in text for w in ["please", "can you", "could you", "review", "confirm"]):
        return "respond", "polite"
    else:
        return "ignore", "neutral"


df[["ideal_intent", "ideal_tone"]] = df["body"].apply(
    lambda x: pd.Series(add_ground_truth(x))
)
```

```python
email_eval_df = df[["id", "body", "ideal_intent", "ideal_tone"]]

email_eval_df.to_csv(
    "../data/email_evaluation_dataset_Nikhitha.csv",
    index=False
)

print("email_evaluation_dataset_Nikhitha.csv created")
```
