```python
import pandas as pd
```

```python
df = pd.read_csv("../data/email_evaluation_dataset_Nikhitha.csv")
df.head()
```

```python
def email_assistant(text):
    t = str(text).lower()
    
    if any(k in t for k in ['invoice','payment','due','meeting','submit','confirm']):
        action = 'respond'
        tone = 'polite'
    elif any(k in t for k in ['password','security','alert','unauthorized','suspicious']):
        action = 'notify'
        tone = 'urgent'
    else:
        action = 'ignore'
        tone = 'neutral'
        
    return action, tone
```

```python
df[['predicted_action','predicted_tone']] = df['email_text'].apply(
    lambda x: pd.Series(email_assistant(x))
)

df.head()
```

```python
df['action_correct'] = df['predicted_action'] == df['expected_action']
df['tone_correct'] = df['predicted_tone'] == df['expected_tone']

df.head()
```

```python
action_accuracy = df['action_correct'].mean() * 100
tone_accuracy = df['tone_correct'].mean() * 100

action_accuracy, tone_accuracy
```

```python
errors = df[df['action_correct'] == False][
    ['email_text','expected_action','predicted_action']
]

errors.head(10)
```

```python
df.to_csv("../data/milestone2_output_Nikhitha.csv", index=False)
```
