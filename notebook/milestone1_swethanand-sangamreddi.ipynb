```python
import pandas as pd
import re
```

```python
df=pd.read_csv("../data/sample_emails_with_triage_200.csv")
df.head()
```

```python
def clean_email(text):
    if isinstance(text, str):
        text = text.lower()
        text = re.sub(r"http\S+", "", text)
        text = re.sub(r"<.*?>", "", text)
        text = re.sub(r"[^a-zA-Z0-9 ]", " ", text)
        text = re.sub(r"\s+", " ", text).strip()
        return text
    else:
        return ""
    
src_col = "email_text" if "email_text" in df.columns else "body"
df["clean_text"] = df[src_col].apply(clean_email)
df.head()
df.head()
```

```python
def triage(text):
    if any(keyword in text for keyword in ["urgent", "immediately", "asap", "action required"]):
        return "respond_or_act"
    elif any(keyword in text for keyword in ["please check", "review", "verify"]):
        return "notify_human"
    else:
        return "ignore"

if "clean_text" not in df.columns:
    src_col = "email_text" if "email_text" in df.columns else "body"
    df["clean_text"] = df[src_col].apply(clean_email)

df["triage"] = df["clean_text"].apply(triage)
df.head()
```

```python
df.to_csv("../data/milestone1_output_swethanand-sangamreddi.csv", index=False)
```
