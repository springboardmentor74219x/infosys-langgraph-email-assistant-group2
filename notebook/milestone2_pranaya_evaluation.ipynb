```python
def triage_rule(text):
    text = text.lower()

    if "refund" in text or "urgent" in text or "password" in text:
        return "notify_human"

    if "newsletter" in text or "promotion" in text:
        return "ignore"

    return "respond_or_act"

df['triage'] = df['clean_text'].apply(triage_rule)
df[['body', 'triage']].head()
```

```python
df['predicted_triage'] = df['clean_text'].apply(triage_rule)
df[['body','predicted_triage']].head()
```

```python
#creating an column of ideal_response
def ideal_response_logic(label):
    if label == 'ignore':
        return 'No action needed'
    elif label == 'respond':
        return 'Send a reply'
    elif label == 'respond_or_act':
        return 'Respond and take required action'
    elif label == 'notify_human':
        return 'Escalate to human'
    else:
        return 'Manual review required'
```

```python
df['ideal_response'] = df['triage_label'].apply(ideal_response_logic)
df[['triage_label', 'ideal_response']].head()
```

```python
#loading the dataset
df = pd.read_csv("data/email_evaluation_dataset_pranaya.csv")
df.head()
```

```python
def email_assistant(email_text):
    text = email_text.lower()
    if "invoice" in text:
        return "respond", "urgent"
    if "meeting" in text:
        return "respond", "polite"
    if "internship" in text:
        return "notify", "polite"
    if "query" in text:
        return "respond", "neutral"
    if "newsletter" in text:
        return "ignore", "neutral"
    return "ignore", "neutral"
```

```python
#tuples splitting into columns
import pandas as pd
df[['predicted_action', 'predicted_tone']] = df['email_text'].apply(
    lambda x: pd.Series(email_assistant(x))
)
```

```python
df[['email_text',
    'expected_action', 'predicted_action',
    'expected_tone', 'predicted_tone']].head()
```

```python
#accuracy calculation
action_accuracy = (df['expected_action'] == df['predicted_action']).mean()*100
tone_accuracy = (df['expected_tone'] == df['predicted_tone']).mean()*100
print(f"Action accuracy (%): {action_accuracy:.2f}")
print(f"Tone accuracy (%): {tone_accuracy:.2f}")
```

```python
#Perform error analysis
action_errors = df[df['expected_action'] != df['predicted_action']]
tone_errors = df[df['expected_tone'] != df['predicted_tone']]
action_errors[
    ['email_text', 'expected_action', 'predicted_action']
]
tone_errors[
    ['email_text', 'expected_tone', 'predicted_tone']
]
```
