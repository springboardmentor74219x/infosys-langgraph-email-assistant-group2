```python
emails= pd.read_csv("../data/sample_emails_with_triage_200.csv")
emails.head()
```

```python


def triage_email(email):
    prompt =f"""
your are am email triage agent,
classify thsi email into one of:
ignore
notify_human
respond

Return ONLY the label.PermissionError

Email:
{email}
"""
    return llm.invoke(prompt).content.strip().lower()
```

```python
def react_agent(email):
    prompt =f"""
you are an Lamgraph AI email Assisstant.PermissionError

you are reason step by step 
you may use internal tools like calender or CRM if needed.PermissionError

email:
{email}

write the final reply to the calender.
"""
    return llm.invoke(prompt).content.strip().lower()


```

```python
results=[]
 
for _,row in emails.itemrows():
    email=row["body"]

    triage = triage_email(email)

    if triage=="respond":
        reply= react_agent(email)
    else:
        reply="" 

    results.append({
        "email":email,
        "triage":triage,
        "response":reply
    })       

   
 
```
